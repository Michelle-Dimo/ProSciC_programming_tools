## First example: the reaction $A \rightarrow B \rightarrow C$

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

# Define the rate constants
k1 = 0.5
k2 = 0.2

# Define the equations
def kinetics(t, concentrations):
    A, B, C = concentrations
    dA_dt = -k1 * A
    dB_dt = k1 * A - k2 * B
    dC_dt = k2 * B
    return [dA_dt, dB_dt, dC_dt]

#Starting concentrations [A0, B0, C0]
y0 = [1.0, 0.0, 0.0]

# Time span for the simulation
t_eval = np.linspace(0, 20, 100)

# Solve the ODEs using SciPy
solution = solve_ivp(kinetics, [0, 20], y0, t_eval=t_eval)

# Plot the results
plt.plot(solution.t, solution.y[0], label='A')
plt.plot(solution.t, solution.y[1], label='B')
plt.plot(solution.t, solution.y[2], label='C')
plt.xlabel('Time')
plt.ylabel('Concentration')
plt.legend()
plt.show()
```

This task is based on the results from PS1 task 8-11, further description in progress

```python
import math 

Tm_kelvin = 333.15
DeltaH_Tm = 400
DeltaC_p = 15


DeltaS_Tm = DeltaH_Tm / Tm_kelvin

T_kelvin = np.linspace(273.15, 373.15, 500)


DeltaH = DeltaH_Tm + DeltaC_p * (T_kelvin - Tm_kelvin)
DeltaS = DeltaS_Tm + DeltaC_p * np.log(T_kelvin/Tm_kelvin)
DeltaG = DeltaH_Tm - T_kelvin*DeltaS_Tm + DeltaC_p*(-T_kelvin*np.log(T_kelvin/Tm_kelvin)+T_kelvin-Tm_kelvin)

plt.figure(figsize=(9, 6))


plt.plot(T_kelvin, DeltaH, label=r'$\Delta H_U(T)$', color='firebrick', lw=2)
plt.plot(T_kelvin, DeltaS, label=r'$\Delta S_U(T)$', color='forestgreen', lw=2)
plt.plot(T_kelvin, T_kelvin*DeltaS, label=r'$T\Delta S_U(T)$', color='yellow', lw=2)
plt.plot(T_kelvin, DeltaG, label=r'$\Delta G_U(T)$', color='royalblue', lw=2.5)

plt.axhline(0, color='black', linestyle='--', alpha=0.5)

plt.title('Folding dynamics of PSaseX', fontsize=14, fontweight='bold')
plt.xlabel('Temperature (°K)', fontsize=12)
plt.ylabel('Energy (kJ / mol)', fontsize=12)
plt.xlim(273.15, 350.15)
plt.ylim(-500,500)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=11, loc='upper right')

# Vis plottet
plt.tight_layout()
plt.show()
```
